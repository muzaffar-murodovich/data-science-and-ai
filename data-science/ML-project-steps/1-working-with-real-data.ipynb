## 5. Balanced train/test set

```python
df['median_income'].hist(bins=50, figsize=(9, 6))
plt.show()
```

```python
df['income_cat'] = pd.cut(df['median_income'], bins=[0., 1.5, 3.0, 4.5, 6.0, np.inf], labels=[1, 2, 3, 4, 5])
df['income_cat'].hist(figsize=(9, 6))
plt.show()
```

```python
df.sample(5)
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
stratified_split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_index, test_index in stratified_split.split(df, df["income_cat"]):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]
```

```python
for _ in (strat_train_set, strat_test_set):
    _.drop('income_cat', axis=1, inplace=True)
```

# 6. Correlation

```python
housing = strat_train_set.copy()
```

```python
housing.plot(kind="scatter", x="longitude", y="latitude")
```

```python
housing.plot(kind="scatter", x="longitude", y="latitude", alpha=0.1)
```

```python
housing.plot(kind="scatter", 
             x="longitude", 
             y="latitude", 
             alpha=0.4,
             s=housing["population"]/50, 
             label="population", 
             figsize=(10,7),
             c="median_house_value", 
             cmap=plt.get_cmap("jet"), 
             colorbar=True,
)
plt.legend()
```

```python
housing.corrwith(housing['median_house_value'], numeric_only=True).sort_values(ascending=False)
```

```python
cols = ['median_house_value','median_income','total_rooms','housing_median_age']
sns.pairplot(housing[cols], height=5)
plt.show()
```
