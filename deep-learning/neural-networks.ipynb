```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
```

```python
X = torch.tensor([[0.2, 0.3],
                  [0.6, 0.8],
                  [0.4, 0.5],
                  [0.7, 0.9]],
                  dtype=torch.float32
                  )
y = torch.tensor([[0], [1], [0], [1]], dtype=torch.float32)
```

```python
model = nn.Sequential(
    nn.Linear(2, 4),
    nn.ReLU(),
    nn.Linear(4, 1),
    nn.Sigmoid()
)
```

```python
loss = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

epochs = 100
losses = []
for epoch in range(epochs):
    y_pred = model(X)
    loss_value = loss(y_pred, y)
    losses.append(loss_value.item())

    optimizer.zero_grad()
    loss_value.backward()
    optimizer.step()

    if epoch % 10 == 0:
        print(f"Epoch: {epoch}, Loss: {loss_value.item()}")
```

```python
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

```python
new_patient = torch.tensor([[0.55, 0.75]], dtype=torch.float32)
y_pred = model(new_patient)
print(y_pred.item())
```
